```python
import os
import sys
import warnings

warnings.filterwarnings("ignore")

project_root = os.path.abspath("..")
src_path = os.path.join(project_root, "src")

sys.path.insert(0, src_path)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

from preprocessing import load_and_preprocess
from external_data.data_fusion import (
    create_fused_dataset,
    get_fused_arrays
)

plt.style.use("seaborn-v0_8")

print("✅ Environment Ready")
```

```python
stages = [
    "Raw",
    "Rolling",
    "Domain",
    "Lag",
    "ROC",
    "Outlier",
    "Weather",
    "Load"
]

feature_counts = [
    5,
    20,
    26,
    38,
    44,
    50,
    55,
    61
]

plt.figure(figsize=(12,5))

bars = plt.bar(
    stages,
    feature_counts,
    edgecolor="black"
)

plt.title("Feature Engineering Progression")
plt.ylabel("Number of Features")

for bar, value in zip(
    bars,
    feature_counts
):
    plt.text(
        bar.get_x()+bar.get_width()/2,
        value+0.3,
        str(value),
        ha="center"
    )

plt.tight_layout()

plt.savefig(
    "../src/feature_progression.png"
)

plt.show()
```

```python
fused_df = create_fused_dataset(
    "../data/ai4i2020.csv"
)

X, y, feature_names = get_fused_arrays(
    fused_df
)

print("Dataset Shape :", fused_df.shape)
print("Feature Count :", len(feature_names))
print("Samples       :", len(fused_df))
print("Failures      :", y.sum())
print(
    "Failure Rate  :",
    round(y.mean()*100,2),
    "%"
)
```

```python
fig = plt.figure(figsize=(14,8))

gs = gridspec.GridSpec(
    2,
    2
)

# --------------------

ax1 = fig.add_subplot(gs[0,0])

categories = [
    "Raw",
    "Rolling",
    "Domain",
    "Lag",
    "ROC",
    "Outlier",
    "Weather",
    "Load"
]

counts = [
    5,
    15,
    6,
    12,
    6,
    6,
    5,
    6
]

ax1.pie(
    counts,
    labels=categories,
    autopct="%1.0f%%"
)

ax1.set_title(
    "Feature Categories"
)

# --------------------

ax2 = fig.add_subplot(gs[0,1])

class_counts = [
    len(y)-y.sum(),
    y.sum()
]

ax2.bar(
    [
        "Normal",
        "Failure"
    ],
    class_counts
)

ax2.set_title(
    "Failure Distribution"
)

# --------------------

ax3 = fig.add_subplot(gs[1,:])

progress = [
    10,
    25,
    40,
    60,
    80,
    100
]

weeks = [
    "W1-D1",
    "W1-D4",
    "W1",
    "W2-D2",
    "W2-D5",
    "W2"
]

ax3.plot(
    weeks,
    progress,
    marker="o"
)

ax3.set_ylim(0,110)

ax3.set_ylabel("Completion %")

ax3.grid(True)

plt.tight_layout()

plt.savefig(
    "../src/week2_final_dashboard.png"
)

plt.show()
```
