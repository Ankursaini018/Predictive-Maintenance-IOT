```python
import warnings
warnings.filterwarnings("ignore")

import os
import sys

PROJECT_ROOT = os.path.abspath("..")
SRC_DIR = os.path.join(PROJECT_ROOT, "src")

if SRC_DIR not in sys.path:
    sys.path.insert(0, SRC_DIR)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_curve,
    roc_auc_score,
    precision_recall_curve,
    average_precision_score,
    classification_report
)

plt.style.use("ggplot")

print("Week 3 Day 4 Environment Ready")
```

```python
from final_model_pipeline import run_final_pipeline

print("Running Final Pipeline...\n")

results = run_final_pipeline()

print("\nPipeline Completed Successfully.")
```

```python
summary = pd.DataFrame({

    "Metric":[

        "Macro F1",

        "ROC AUC",

        "Precision",

        "Recall"

    ],

    "Score":[

        results["mean_f1"],

        results["mean_auc"],

        results["mean_precision"],

        results["mean_recall"]

    ]

})

summary
```

```python
cm = confusion_matrix(
    results["all_y_true"],
    results["all_y_pred"]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "No Failure",
        "Failure"
    ]
)

fig, ax = plt.subplots(figsize=(7,6))

disp.plot(ax=ax)

plt.title("Final Confusion Matrix")

plt.show()
```

```python
fpr, tpr, _ = roc_curve(
    results["all_y_true"],
    results["all_y_proba"]
)

auc = roc_auc_score(
    results["all_y_true"],
    results["all_y_proba"]
)

plt.figure(figsize=(8,6))

plt.plot(
    fpr,
    tpr,
    label=f"AUC = {auc:.4f}"
)

plt.plot(
    [0,1],
    [0,1],
    "--"
)

plt.xlabel("False Positive Rate")

plt.ylabel("True Positive Rate")

plt.title("ROC Curve")

plt.legend()

plt.grid(True)

plt.show()
```

```python
precision, recall, _ = precision_recall_curve(

    results["all_y_true"],

    results["all_y_proba"]

)

ap = average_precision_score(

    results["all_y_true"],

    results["all_y_proba"]

)

plt.figure(figsize=(8,6))

plt.plot(

    recall,

    precision,

    label=f"AP = {ap:.4f}"

)

plt.xlabel("Recall")

plt.ylabel("Precision")

plt.title("Precision-Recall Curve")

plt.legend()

plt.grid(True)

plt.show()
```

```python
report = classification_report(

    results["all_y_true"],

    results["all_y_pred"],

    output_dict=True,

    zero_division=0

)

report_df = pd.DataFrame(report).T

report_df
```

```python
weeks = [

    "Week 1",

    "Week 2",

    "Week 3"

]

scores = [

    0.65,

    0.80,

    results["mean_f1"]

]

plt.figure(figsize=(8,5))

bars = plt.bar(

    weeks,

    scores

)

plt.axhline(

    0.85,

    linestyle="--",

    label="Target"

)

for bar, score in zip(bars, scores):

    plt.text(

        bar.get_x()+bar.get_width()/2,

        score+0.01,

        f"{score:.3f}",

        ha="center"

    )

plt.title("Project Journey")

plt.ylabel("Macro F1")

plt.legend()

plt.show()
```
