```python
import warnings
warnings.filterwarnings("ignore")

import os
import sys

PROJECT_ROOT = os.path.abspath("..")
SRC_DIR = os.path.join(PROJECT_ROOT, "src")

if SRC_DIR not in sys.path:
    sys.path.insert(0, SRC_DIR)

import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("ggplot")

print("Notebook Environment Ready")
```

```python
results_path = os.path.join(
    PROJECT_ROOT,
    "src",
    "tuning_results",
    "robustness_results.csv"
)

results = pd.read_csv(results_path)

print(results.shape)

results.head()
```

```python
results.describe()
```

```python
plt.figure(figsize=(10,6))

for noise in results["noise_type"].unique():

    if noise == "clean":
        continue

    subset = results[
        results["noise_type"] == noise
    ]

    plt.plot(

        subset["noise_level"],

        subset["macro_f1"],

        marker="o",

        linewidth=2,

        label=noise.capitalize()

    )

plt.xlabel("Noise Level")

plt.ylabel("Macro F1")

plt.title("Model Robustness")

plt.grid(True)

plt.legend()

plt.show()
```

```python
plt.figure(figsize=(10,6))

for noise in results["noise_type"].unique():

    if noise == "clean":
        continue

    subset = results[
        results["noise_type"] == noise
    ]

    plt.plot(

        subset["noise_level"],

        subset["performance_drop"],

        marker="o",

        linewidth=2,

        label=noise.capitalize()

    )

plt.xlabel("Noise Level")

plt.ylabel("Performance Drop")

plt.title("Performance Degradation")

plt.grid(True)

plt.legend()

plt.show()
```

```python
summary = (

    results[
        results["noise_type"] != "clean"
    ]

    .groupby("noise_type")

    .agg(

        Mean_F1=("macro_f1","mean"),

        Worst_F1=("macro_f1","min"),

        Maximum_Drop=("performance_drop","max")

    )

)

summary
```

```python
worst = results.loc[
    results["macro_f1"].idxmin()
]

worst
```
