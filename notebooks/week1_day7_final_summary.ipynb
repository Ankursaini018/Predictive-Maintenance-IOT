```python
"""
Week 1 - Day 7
Final Summary Notebook

Contextual Predictive Maintenance
Infotact DS/ML Internship
"""

import os
import sys
import warnings

warnings.filterwarnings("ignore")

project_root = os.path.abspath("..")
src_path = os.path.join(project_root, "src")

sys.path.insert(0, src_path)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from preprocessing import load_and_preprocess

plt.style.use("seaborn-v0_8")

print("Environment Ready")
```

```python
X, y, feature_names, df = load_and_preprocess(
    "../data/ai4i2020.csv"
)

print("="*60)
print("WEEK 1 FINAL SUMMARY")
print("="*60)

print(f"Dataset Shape      : {df.shape}")
print(f"Total Features     : {len(feature_names)}")
print(f"Failure Samples    : {int(y.sum())}")
print(f"Failure Rate       : {y.mean()*100:.2f}%")
```

```python
summary = {
    "Rolling":
        len([c for c in feature_names if "roll" in c]),

    "Lag":
        len([c for c in feature_names if "lag" in c]),

    "ROC":
        len([c for c in feature_names if "roc" in c]),

    "Outlier":
        len([c for c in feature_names if "outlier" in c]),

    "Domain":
        len([
            c for c in feature_names
            if c in [
                "power",
                "temp_delta",
                "tool_wear_rate",
                "torque_per_rpm",
                "temp_wear_interaction"
            ]
        ])
}

summary_df = pd.DataFrame(
    summary.items(),
    columns=[
        "Feature Category",
        "Count"
    ]
)

display(summary_df)
```

```python
# Failure Distribution

plt.figure(figsize=(6, 5))

# Convert numpy array to pandas Series
counts = pd.Series(y).value_counts().sort_index()

plt.bar(
    ["Normal", "Failure"],
    counts.values,
    edgecolor="black",
    color=["steelblue", "tomato"]
)

plt.title("Failure Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Samples")

# Add value labels on bars
for i, value in enumerate(counts.values):
    plt.text(
        i,
        value + max(counts.values) * 0.01,
        str(value),
        ha="center",
        fontsize=10
    )

plt.tight_layout()

plt.savefig(
    "../src/week1_final_failure_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Class Distribution:")
print(counts)

print(f"\nFailure Rate: {counts[1] / counts.sum() * 100:.2f}%")
```

```python
corr = (
    pd.DataFrame(
        X,
        columns=feature_names
    )
    .assign(**{
        "Machine failure":y
    })
    .corr()["Machine failure"]
    .drop("Machine failure")
    .sort_values(
        key=abs,
        ascending=False
    )
)

top10 = corr.head(10)

plt.figure(figsize=(10,6))

top10.sort_values().plot(
    kind="barh",
    edgecolor="black"
)

plt.title(
    "Top Features Correlated with Failure"
)

plt.tight_layout()

plt.savefig(
    "../src/week1_top10_features.png"
)

plt.show()

display(top10)
```
