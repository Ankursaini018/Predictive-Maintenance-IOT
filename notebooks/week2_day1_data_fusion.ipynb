```python
"""
Week 2 - Day 1
Contextual Data Fusion Notebook
================================
Merging IoT sensor data with external
weather and factory load context
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys

sys.path.append('../src')

from external_data.data_fusion import (
    create_fused_dataset,
    get_fused_arrays
)

plt.style.use('seaborn-v0_8')

# Run full fusion pipeline
fused_df = create_fused_dataset(
    '../data/ai4i2020.csv'
)

X, y, feature_names = (
    get_fused_arrays(
        fused_df
    )
)

print(f"\n✅ Fused dataset ready!")
print(f"   Shape    : {fused_df.shape}")
print(f"   Features : {len(feature_names)}")
```

```python
# External contextual features

external_features = [
    'ambient_temp_c',
    'humidity_pct',
    'wind_speed_kmh',
    'air_pressure_hpa',
    'weather_encoded',
    'factory_load',
    'load_encoded'
]

print("╔══════════════════════════════╗")
print("║ EXTERNAL CONTEXT FEATURES   ║")
print("╠══════════════════════════════╣")

for feature in external_features:
    print(f"║ {feature:<28} ║")

print("╠══════════════════════════════╣")
print(
    f"║ TOTAL : {len(external_features):<20} ║"
)
print("╚══════════════════════════════╝")
```

```python
# Compare failures against contextual features

context_cols = [
    'ambient_temp_c',
    'humidity_pct',
    'factory_load'
]

fig, axes = plt.subplots(
    1,
    3,
    figsize=(18, 5)
)

for i, col in enumerate(context_cols):

    sns.boxplot(
        data=fused_df,
        x='Machine failure',
        y=col,
        ax=axes[i]
    )

    axes[i].set_title(
        f'{col} vs Failure',
        fontsize=12,
        fontweight='bold'
    )

plt.tight_layout()
plt.show()
```

```python
# Correlation with machine failure

corr = (
    fused_df[
        external_features +
        ['Machine failure']
    ]
    .corr()['Machine failure']
    .drop('Machine failure')
    .sort_values(
        key=abs,
        ascending=False
    )
)

plt.figure(figsize=(8, 5))

corr.plot(
    kind='barh'
)

plt.title(
    'External Features Correlated with Failure',
    fontsize=13,
    fontweight='bold'
)

plt.xlabel('Correlation')
plt.axvline(
    0,
    color='black',
    linewidth=0.8
)

plt.tight_layout()
plt.show()

print("\n📊 External Feature Correlations:")
print(corr)
```
