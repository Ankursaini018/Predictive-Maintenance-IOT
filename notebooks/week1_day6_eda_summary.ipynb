```python
"""
Week 1 - Day 6
EDA Summary Notebook

Consolidating all findings from Week 1
"""

import os
import sys
import warnings

warnings.filterwarnings("ignore")

project_root = os.path.abspath("..")
src_path = os.path.join(project_root, "src")

sys.path.insert(0, src_path)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from preprocessing import load_and_preprocess

plt.style.use("seaborn-v0_8")

print("Loading processed dataset...")

X, y, feature_names, df = load_and_preprocess(
    "../data/ai4i2020.csv"
)

print("\nDataset Loaded Successfully")
print("-"*40)
print(f"Shape            : {df.shape}")
print(f"Total Features   : {len(feature_names)}")
print(f"Failure Samples  : {int(y.sum())}")
print(f"Failure Rate     : {y.mean()*100:.2f}%")
```

```python
print("="*60)
print("DATASET SUMMARY")
print("="*60)

summary = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str),
    "Missing Values": df.isnull().sum().values
})

display(summary)

print("\nTotal Missing Values :", df.isnull().sum().sum())
print("Duplicate Rows       :", df.duplicated().sum())
```

```python
feature_stats = pd.DataFrame({
    "Feature": feature_names,
    "Mean": X.mean(axis=0).round(3),
    "Std": X.std(axis=0).round(3),
    "Min": X.min(axis=0).round(3),
    "Max": X.max(axis=0).round(3)
})

print("="*60)
print("FEATURE STATISTICS")
print("="*60)

display(feature_stats)
```

```python
corr = (
    pd.DataFrame(X, columns=feature_names)
    .assign(**{"Machine failure": y})
    .corr()["Machine failure"]
    .drop("Machine failure")
    .sort_values(key=abs, ascending=False)
)

top10 = corr.head(10)

plt.figure(figsize=(10,6))

top10.sort_values().plot(
    kind="barh",
    edgecolor="black"
)

plt.title("Top 10 Features Correlated with Failure")

plt.tight_layout()

plt.savefig(
    "../src/eda_top10_correlation.png"
)

plt.show()

print(top10)
```

```python
summary = {
    "Rolling Features":
        len([c for c in feature_names if "roll" in c]),

    "Lag Features":
        len([c for c in feature_names if "lag" in c]),

    "ROC Features":
        len([c for c in feature_names if "roc" in c]),

    "Outlier Features":
        len([c for c in feature_names if "outlier" in c]),

    "Domain Features":
        len([
            c for c in feature_names
            if c in [
                "power",
                "temp_delta",
                "tool_wear_rate",
                "torque_per_rpm",
                "temp_wear_interaction"
            ]
        ])
}

summary_df = pd.DataFrame(
    summary.items(),
    columns=[
        "Feature Type",
        "Count"
    ]
)

display(summary_df)
```
