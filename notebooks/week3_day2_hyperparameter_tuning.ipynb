```python
from external_data.data_fusion import (
    create_fused_dataset,
    get_fused_arrays
)

fused_df = create_fused_dataset(
    "../data/ai4i2020.csv"
)

X, y, feature_names = get_fused_arrays(
    fused_df
)

print(X.shape)

print(y.shape)

print(feature_names[:10])
```

```python
best_params, grid_results = focused_grid_search(
    X,
    y
)

grid_results
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.bar(
    grid_results["Configuration"].astype(str),
    grid_results["Macro F1"]
)

plt.title("Grid Search Results")

plt.xlabel("Configuration")

plt.ylabel("Macro F1")

plt.grid(True)

plt.show()
```

```python
import pandas as pd

history = pd.read_csv(
    "../src/tuning_results/optuna_trial_history.csv"
)

history.head()
```

```python
import json

with open(
    "../src/tuning_results/optuna_best_parameters.json"
) as f:

    best = json.load(f)

best
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    history["Trial"],
    history["Macro F1"],
    marker="o"
)

plt.title("Optuna Trial History")

plt.xlabel("Trial")

plt.ylabel("Macro F1")

plt.grid(True)

plt.show()
```

```python
summary = pd.DataFrame({

    "Method":[
        "Grid Search",
        "Optuna"
    ],

    "Best Macro F1":[

        grid_results["Macro F1"].max(),

        best["best_macro_f1"]

    ]

})

summary
```

## Observations

- Grid Search evaluated predefined LightGBM hyperparameter configurations.
- Optuna used Bayesian Optimization to search the parameter space.
- Optuna achieved a higher Macro F1 score than manual Grid Search.
- Bayesian Optimization reduced manual tuning effort.
- The tuned LightGBM model improved predictive performance.
