```python
import warnings
warnings.filterwarnings("ignore")

import os
import sys

PROJECT_ROOT = os.path.abspath("..")
SRC_DIR = os.path.join(PROJECT_ROOT, "src")

if SRC_DIR not in sys.path:
    sys.path.insert(0, SRC_DIR)

import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("ggplot")

print("Notebook Environment Ready")
```

```python
from shap_visualizer import (
    load_shap_ranking,
    categorize_features
)

print("SHAP modules imported successfully.")
```

```python
df = load_shap_ranking()

df = categorize_features(df)

df.head()
```

```python
print("="*60)
print("Top 20 Most Important Features")
print("="*60)

df.head(20)
```

```python
plt.figure(figsize=(10,7))

plt.barh(
    df.head(20)["Feature"][::-1],
    df.head(20)["Mean SHAP"][::-1]
)

plt.title("Top 20 SHAP Features")

plt.xlabel("Mean SHAP")

plt.tight_layout()

plt.show()
```

```python
category_importance = (

    df.groupby("Category")["Mean SHAP"]

    .sum()

    .sort_values(ascending=False)

)

category_importance
```

```python
plt.figure(figsize=(8,8))

plt.pie(

    category_importance,

    labels=category_importance.index,

    autopct="%1.1f%%",

    startangle=90

)

plt.title("Feature Category Contribution")

plt.show()
```

```python
plt.figure(figsize=(10,6))

cumulative = (

    df["Mean SHAP"]

    .cumsum()

    / df["Mean SHAP"].sum()

) * 100

plt.plot(

    range(1, len(df)+1),

    cumulative,

    marker="o"

)

plt.grid(True)

plt.title("Cumulative SHAP Importance")

plt.xlabel("Features")

plt.ylabel("Importance (%)")

plt.show()
```

```python
summary = pd.DataFrame({

    "Metric":[

        "Total Features",

        "Top Feature",

        "Top Feature Importance",

        "Categories"

    ],

    "Value":[

        len(df),

        df.iloc[0]["Feature"],

        round(df.iloc[0]["Mean SHAP"],4),

        df["Category"].nunique()

    ]

})

summary
```

## Observations

- SHAP provides local and global explanations for model predictions.
- The top-ranked features contribute most to machine failure prediction.
- Category-level aggregation helps interpret groups of related features.
- Feature importance decreases rapidly after the top-ranked variables.
- SHAP improves model transparency and interpretability.
